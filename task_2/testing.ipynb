```python
from ultralytics import YOLO
import matplotlib.pyplot as plt

model = YOLO('yolo26s.pt')
model.train(data='./dataset/data.yaml', epochs = 50)
```

```python
import cv2
import zxingcpp as zx


cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Не удалось открыть камеру")

while True:
    # Читаем очередной кадр
    ret, frame = cap.read()
    x, y = frame.shape[0], frame.shape[1]
    # frame = frame[x//3:2*x//3, y//3 : 2*y//3]
    frame = cv2.resize(frame, None, fx = 0.5, fy = 0.5)

    if not ret:
        print("Не удалось получить кадр")
        break
    
    frame = model.predict(frame)[0].plot()
    # barcodes = zx.read_barcodes(frame)
    # barcode_string = ' '.join([barcode.text for barcode in barcodes])
    # print(barcodes)
    # frame = cv2.putText(frame, barcode_string, (frame.shape[1]//2, frame.shape[0]//2), 
    #                     fontFace=cv2.FONT_HERSHEY_PLAIN, color = (0, 0, 0), fontScale=1, )
    # Выводим кадр в отдельное окно
    cv2.imshow("Camera", frame)

    # Ждём 1 мс и проверяем нажатие клавиши
    # q — выход
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Освобождаем камеру
cap.release()

# Закрываем окна OpenCV
cv2.destroyAllWindows()
```
